Code Reuse Session 1 - Code block 22

```python
#function to detect outliers in any variable based on the interquartile range
def find_outliers_IQR(data):
  q1=data.quantile(0.25)
  q3=data.quantile(0.75)
  IQR=q3-q1
  outliers = data[((data<(q1-1.5*IQR))|(data> (q3+1.5*IQR)))]
  return outliers
```
