Code Reuse Session 2 - code block 6

```python
# Drop 'Mortality_Status' column to get feature set
X = data.drop(['Mortality_Status'], axis=1)
# Select 'Mortality_Status' column as the target variable
y = data['Mortality_Status']
```

Code Reuse Session 2 - code block 7

```python
# to split data into training and testing sets
from sklearn.model_selection import train_test_split
```

```python
# Split data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=14, stratify=y)
```

Code Reuse Session 2 - code block 11

```python
# Fit the model on training data
logreg.fit(X_train, y_train)
```

Code Reuse Session 2 - code block 12

```python
# Predict on training data
y_pred_train = logreg.predict(X_train)
```

Code Reuse Session 2 - code block 13

```python
# Predict on test data
y_pred = logreg.predict(X_test)
```

```python
# Import GridSearchCV for hyperparameter tuning
from sklearn.model_selection import GridSearchCV
```

Code Reuse Session 2 - code block 26

```python
# Create a new KNN model
knn = KNeighborsClassifier()

# Create a dictionary of hyperparameters (n_neighbors and distance metrics) to test
param_grid = {'n_neighbors': range(1, 25), 'metric': ['euclidean', 'manhattan']}

# Use GridSearchCV to test different hyperparameters with cross-validation and ROC AUC scoring
knn_gscv = GridSearchCV(knn, param_grid, cv=5, scoring='roc_auc')

# Fit the grid search model on the data
knn_gscv.fit(X, y)
```

Code Reuse Session 2 - code block 27

```python
# Check the top performing n_neighbors value and best hyperparameters
knn_gscv.best_params_
```

Code Reuse Session 2 - code block 28

```python
# Perform prediction on the test dataset using the best KNN model
y_pred = knn_gscv.predict(X_test)
# Construct a confusion matrix to compare actual vs predicted values for KNN model
cm_knn_gscv = confusion_matrix(y_test, y_pred, labels=knn_gscv.classes_)
# Display the confusion matrix for the KNN model
disp_knn_gscv_cm = ConfusionMatrixDisplay(cm_knn_gscv, display_labels=knn_gscv.classes_)
# Plot the confusion matrix for the KNN model
disp_knn_gscv_cm.plot()
# Display the classification report (precision, recall, f1-score) for KNN model
print(classification_report(y_test, y_pred))
```
